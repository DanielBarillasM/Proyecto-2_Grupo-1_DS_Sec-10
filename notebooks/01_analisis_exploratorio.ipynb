## 1. Situación problemática

Cuando un estudiante de primaria o secundaria resuelve mal un problema de matemáticas, la respuesta
incorrecta es apenas el síntoma. Lo que realmente le sirve a un docente es el *porqué*: el razonamiento
defectuoso que produjo esa respuesta. En la literatura educativa a ese razonamiento defectuoso se le
llama **misconception** (concepción errónea).

Un ejemplo concreto del propio dataset: ante la pregunta *"¿qué fracción de la figura no está sombreada?"*,
dos estudiantes pueden responder ambos $\frac{3}{6}$, pero por motivos distintos — uno contó mal las
partes, otro invirtió numerador y denominador. Son errores **distintos** que requieren
intervenciones pedagógicas **distintas**, aunque la respuesta marcada sea idéntica.

El problema práctico es de **escala**. Diagnosticar una concepción errónea exige leer la explicación
escrita del estudiante e interpretarla. Un docente con 150 estudiantes no puede hacer eso de forma
sistemática después de cada evaluación. Las concepciones erróneas, si no se detectan, son
persistentes: se acumulan y bloquean el aprendizaje de temas posteriores que dependen de ellas.

Este reto — organizado por **Vanderbilt University** y **The Learning Agency**, con datos de la
plataforma **Eedi** — busca automatizar ese diagnóstico: dado lo que el estudiante escribió,
predecir qué tipo de error cometió.

<a id="2"></a>
## 2. Problema científico

> **¿Es posible predecir automáticamente, a partir de la explicación en lenguaje natural escrita por un
> estudiante junto con la opción de respuesta que seleccionó, si su razonamiento es correcto y —cuando
> no lo es— qué concepción errónea específica lo explica?**

Formalmente es un problema de **clasificación multiclase sobre texto corto**. Para cada observación se
dispone del enunciado de la pregunta, la respuesta de opción múltiple marcada y la explicación libre del
estudiante; se debe predecir la etiqueta compuesta `Category:Misconception`.

Las dificultades que anticipamos y que este análisis busca cuantificar son:

- **Texto ruidoso y muy corto.** Son estudiantes de 4.º a 8.º grado; hay errores ortográficos, ausencia
  de mayúsculas y notación matemática improvisada.
- **Desbalance severo de clases.** Las concepciones erróneas raras tienen muy pocos ejemplos.
- **Dependencia del contexto.** Una misma frase puede indicar razonamiento correcto en una pregunta y
  erróneo en otra.

## 5. Limpieza y preprocesamiento

### 5.1 Diagnóstico de los valores faltantes

El 73.1% de nulos en `Misconception` es la primera señal de alarma de cualquier EDA. Antes de decidir
imputar o descartar, hay que entender **por qué** faltan.

````python
# Category tiene forma Veracidad_Tipo -> la descomponemos
train["Veracidad"] = train["Category"].str.split("_").str[0]   # True / False
train["Tipo"] = train["Category"].str.split("_").str[1]        # Correct / Misconception / Neither

tab = pd.crosstab(train["Tipo"], train["Misconception"].notna(),
                  rownames=["Tipo de razonamiento"], colnames=["¿Tiene Misconception?"])
tab.columns = ["No (nulo)", "Sí"]
print(tab.to_string())

coincide = (train["Misconception"].notna() == train["Tipo"].eq("Misconception")).all()
print(f"\n¿'Misconception' no-nulo ocurre EXACTAMENTE cuando Tipo == 'Misconception'? -> {coincide}")
````

**Hallazgo 1 — Los nulos son estructurales, no faltantes.**

La correspondencia es perfecta: `Misconception` tiene valor si y solo si `Tipo == "Misconception"`.
No hay un solo caso de dato perdido. El nulo significa *"no aplica: este razonamiento no contiene una
concepción errónea"*.

**Decisión de limpieza:** **no se imputa ni se elimina nada.** Se reemplaza el nulo por la categoría
explícita `"NA"`, que es además la convención que exige el formato de submission. Imputar aquí
(con la moda, por ejemplo) habría introducido ruido masivo en el 73% de los datos — un error grave
y frecuente cuando no se diagnostica el origen de la ausencia.

````python
df = train.copy()
df["Misconception"] = df["Misconception"].fillna("NA")

# Etiqueta objetivo compuesta
df["target"] = df["Category"] + ":" + df["Misconception"]

print("Nulos restantes en todo el dataset:", df.isna().sum().sum())
print("Cardinalidad de la etiqueta compuesta:", df["target"].nunique(), "clases")
````

### 5.2 Duplicados

**Hallazgo 2 — No hay filas completamente idénticas, pero sí repeticiones que deben agruparse.**

No existen filas idénticas cuando se conserva `row_id`; al excluirlo aparecen 736 registros iguales y

pueden ser respuestas válidas; deben mantenerse juntas dentro del mismo pliegue de validación para
evitar fuga de información.

### 5.3 Calidad del texto

Antes de limpiar, medimos el estado real del texto para justificar cada transformación.

````python
s = df["StudentExplanation"]
calidad = pd.Series({
    "Empiezan en minúscula":        s.str[0].str.islower().mean(),
    "Contienen dígitos":            s.str.contains(r"\d", regex=True).mean(),
    "Contienen '/' (fracciones)":   s.str.contains("/", regex=False).mean(),
    "Contienen '=' ":               s.str.contains("=", regex=False).mean(),
    "Menos de 3 palabras":          s.str.split().str.len().lt(3).mean(),
    "Contienen notación LaTeX":     s.str.contains(r"\\\(", regex=True).mean(),
}) * 100

print("CALIDAD DEL TEXTO — StudentExplanation (% de observaciones)")
print("-" * 58)
for k, v in calidad.items():
    print(f"{k:<32} {v:>6.1f}%")

print("\nOtras columnas:")
print(f"{'MC_Answer con notación LaTeX':<32} {df['MC_Answer'].str.contains(r'\\\(').mean()*100:>6.1f}%")
print(f"{'QuestionText con [Image: ...]':<32} {df['QuestionText'].str.contains(r'\[Image').mean()*100:>6.1f}%")
````

**Hallazgo 3 — El texto es informal pero no está corrupto.**

- El **62.8%** empieza en minúscula y hay faltas de ortografía frecuentes (*"0ne third is equal to tree
  nineth"*). Es escritura auténtica de estudiantes de 4.º a 8.º grado.
- El **78.3%** contiene dígitos y el **27.2%** el símbolo `/`. El contenido matemático está en el texto,
  no solo en palabras. **Implicación directa:** un tokenizador que elimine números —práctica estándar en
  NLP— destruiría la señal más importante. **No se eliminan dígitos ni símbolos matemáticos.**
- `StudentExplanation` **no** contiene LaTeX (0.0%), pero `MC_Answer` **sí** en el 96% de los casos
  (`\( \frac{1}{3} \)`). Son dos convenciones de escritura distintas que conviene normalizar por
  separado.
- Solo 3 de las 15 preguntas incluyen `[Image: ...]`, una descripción textual de una figura. La imagen
  real no está disponible; queda como texto.

````python
def limpiar_latex(t: str) -> str:
    # Convierte la notacion LaTeX de MC_Answer a texto plano legible
    t = str(t)
    t = re.sub(r"\\frac\{([^}]*)\}\{([^}]*)\}", r"\1/\2", t)  # \frac{1}{3} -> 1/3
    t = re.sub(r"\\[a-zA-Z]+", " ", t)                        # otros comandos
    t = t.replace("\\(", " ").replace("\\)", " ")
    t = re.sub(r"[{}$]", " ", t)
    return re.sub(r"\s+", " ", t).strip()

df["MC_Answer_clean"] = df["MC_Answer"].apply(limpiar_latex)
df["Explanation_clean"] = (df["StudentExplanation"].str.lower()
                                                   .str.replace(r"\s+", " ", regex=True)
                                                   .str.strip())

print("Ejemplos de normalización de MC_Answer:")
print(df[["MC_Answer", "MC_Answer_clean"]].drop_duplicates().head(6).to_string(index=False))
````

**Resumen de decisiones de limpieza**

| # | Situación detectada | Decisión | Justificación |
|---|---|---|---|
| 1 | 73.1% de nulos en `Misconception` | Reemplazar por `"NA"` | Son estructurales; imputar corrompería el 73% |
| 2 | 0 filas duplicadas | Ninguna acción | No hay redundancia |
| 3 | 970 textos repetidos | Conservar | Observaciones independientes y legítimas |
| 4 | LaTeX en 96% de `MC_Answer` | Convertir a texto plano | Homogeneiza el formato numérico |
| 5 | 78% del texto con dígitos | **No eliminar números** | Son la señal matemática principal |
| 6 | Mayúsculas inconsistentes | Minusculizar `Explanation` | Reduce dispersión del vocabulario |

### 6.2 Variables cuantitativas derivadas

Como se estableció en la sección 4.3, el dataset **no tiene variables cuantitativas nativas**. Para
poder aplicar estadística descriptiva, correlaciones y detección de outliers —y porque en NLP estas
características son predictoras genuinas— derivamos variables numéricas del texto.

````python
e = df["Explanation_clean"]

df["n_chars"]     = e.str.len()
df["n_words"]     = e.str.split().str.len()
df["n_unique_w"]  = e.str.split().apply(lambda w: len(set(w)))
tokens = e.str.findall(r"\S+")
df["avg_token_len"] = tokens.apply(lambda ts: np.mean([len(t) for t in ts]) if ts else 0.0)
df["lex_diversity"] = df["n_unique_w"] / df["n_words"]
df["n_digits"]    = e.str.count(r"\d")
df["digit_ratio"] = df["n_digits"] / df["n_chars"]
df["n_math_sym"]  = e.str.count(r"[+\-*/=<>%]")
df["has_fraction"] = e.str.contains(r"\d\s*/\s*\d", regex=True).astype(int)
def contar_oraciones(texto):
    partes = re.split(r"[.!?]+(?:\s+|$)", texto)
    return max(1, sum(bool(parte.strip()) for parte in partes))

df["n_sentences"] = e.apply(contar_oraciones)

NUM = ["n_chars","n_words","n_unique_w","avg_token_len","lex_diversity",
       "n_digits","digit_ratio","n_math_sym","n_sentences"]

print("RESUMEN ESTADÍSTICO — variables cuantitativas derivadas")
resumen = df[NUM].describe().T
resumen["IQR"] = resumen["75%"] - resumen["25%"]
resumen["CV"] = (resumen["std"] / resumen["mean"]).round(2)
resumen["asimetría"] = df[NUM].skew().round(2)
resumen.round(2)
````

**Lectura del resumen:** todas las variables de longitud presentan **asimetría positiva** fuerte
(`n_chars` ≈ 2.4, `n_words` ≈ 2.3): muchas explicaciones cortas y una cola de explicaciones largas.
La mediana de `n_words` es **13 palabras** frente a un máximo de **177**. Son textos muy cortos, lo que
limita la utilidad de modelos que dependen de contexto largo.

````python
fig, axes = plt.subplots(2, 3, figsize=(16, 8))
vars_plot = ["n_words", "n_chars", "avg_token_len", "lex_diversity", "digit_ratio", "n_math_sym"]

for ax, v in zip(axes.flat, vars_plot):
    sns.histplot(df[v], bins=45, ax=ax, color="#2980b9", edgecolor="white", linewidth=.3)
    ax.axvline(df[v].mean(), color="red", ls="--", lw=1.4, label=f"media {df[v].mean():.2f}")
    ax.axvline(df[v].median(), color="orange", ls="-", lw=1.4, label=f"mediana {df[v].median():.2f}")
    ax.set_title(f"Histograma de {v}"); ax.legend(fontsize=8); ax.set_ylabel("")

plt.suptitle("Distribución de las variables cuantitativas derivadas", fontweight="bold", y=1.01)
plt.tight_layout(); plt.show()
````

### 6.3 Análisis de outliers

````python
def outliers_iqr(serie):
    q1, q3 = serie.quantile([.25, .75])
    iqr = q3 - q1
    lo, hi = q1 - 1.5*iqr, q3 + 1.5*iqr
    m = (serie < lo) | (serie > hi)
    return m, lo, hi

res = []
for v in NUM:
    m, lo, hi = outliers_iqr(df[v])
    res.append({"Variable": v, "Límite inf.": round(lo,2), "Límite sup.": round(hi,2),
                "N.º outliers": int(m.sum()), "% outliers": round(m.mean()*100, 2)})
print("DETECCIÓN DE OUTLIERS — método del rango intercuartílico (1.5·IQR)")
print(pd.DataFrame(res).to_string(index=False))
````

````python
m_w, _, hi_w = outliers_iqr(df["n_words"])

print(f"Umbral superior para n_words: {hi_w:.0f} palabras\n")
print("=== Explicaciones ANORMALMENTE LARGAS (top 3) ===")
for _, r in df.nlargest(3, "n_words")[["n_words","Category","StudentExplanation"]].iterrows():
    print(f"\n[{r.n_words} palabras · {r.Category}]\n  {r.StudentExplanation[:260]}...")

print("\n\n=== Explicaciones ANORMALMENTE CORTAS (<= 2 palabras) ===")
cortas = df[df["n_words"] <= 2]
print(f"Total: {len(cortas)} observaciones ({len(cortas)/len(df)*100:.2f}%)")
print(cortas[["n_words","Category","StudentExplanation"]].head(8).to_string(index=False))
````

**Hallazgo 7 — Los outliers son datos válidos, no errores de captura.**

Aproximadamente el **4-5%** de las observaciones son outliers por longitud según el criterio IQR. Al
inspeccionarlos manualmente:

- **Las largas** son estudiantes que explican su procedimiento paso a paso. Texto legítimo y de alta
  calidad informativa.
- **Las cortas** (`"3/9"`, `"-13"`, `"times by"`) son respuestas mínimas: el estudiante escribió solo
  una operación o no se esforzó. También son legítimas — y de hecho son **señal predictiva** de la
  categoría `Neither`.

**Decisión: no se eliminan.** En datos de texto generado por humanos, la longitud atípica es
información sobre el comportamiento del estudiante, no un defecto de medición. Eliminarlos sesgaría el
modelo justo en los casos que más importa detectar. Sí conviene, en la fase de modelado, aplicar una
transformación logarítmica a las variables de longitud para reducir su asimetría.

- Si se particiona **agrupando por `QuestionId`** (`GroupKFold`), el 91% de las concepciones erróneas

propone la estrategia a seguir.

> **Corrección respecto a la práctica habitual:** la recomendación automática de *"agrupar siempre por

### 6.6 Contenido léxico: ¿qué palabras distinguen a cada categoría?

````python
from sklearn.feature_extraction.text import TfidfVectorizer

STOP = {"the","a","an","is","it","to","and","of","i","you","that","this","because","so",
        "in","on","for","be","was","as","are","if","its","my","he","she","they","we"}

vec = TfidfVectorizer(max_features=3000, ngram_range=(1,2), min_df=5, stop_words=list(STOP))
X = vec.fit_transform(df["Explanation_clean"])
vocab = np.array(vec.get_feature_names_out())

print(f"Matriz TF-IDF: {X.shape[0]:,} x {X.shape[1]:,}")
print(f"Densidad: {X.nnz / (X.shape[0]*X.shape[1]) * 100:.3f}% (altamente dispersa)\n")

print("TÉRMINOS MÁS CARACTERÍSTICOS POR TIPO DE RAZONAMIENTO")
print("=" * 72)
for tipo in ["Correct", "Misconception", "Neither"]:
    m = (df["Tipo"] == tipo).values
    dentro = np.asarray(X[m].mean(axis=0)).ravel()
    fuera = np.asarray(X[~m].mean(axis=0)).ravel()
    top = vocab[np.argsort(dentro - fuera)[::-1][:12]]
    print(f"\n{tipo:<14} ({m.sum():,} obs.)")
    print("   " + " · ".join(top))
````

````python
fig, ax = plt.subplots(figsize=(12, 5.5))

palabras = ["divide","multiply","add","subtract","half","equal","times","fraction",
            "negative","guess","know","think"]
filas = []
for p in palabras:
    tiene = df["Explanation_clean"].str.contains(rf"\b{p}", regex=True)
    if tiene.sum() > 40:
        for tipo in ["Correct","Misconception","Neither"]:
            filas.append({"Palabra": p, "Tipo": tipo,
                          "% con la palabra": (df.loc[tiene,"Tipo"] == tipo).mean()*100})

pv = pd.DataFrame(filas).pivot(index="Palabra", columns="Tipo", values="% con la palabra")
pv = pv[["Correct","Misconception","Neither"]].sort_values("Correct")
pv.plot(kind="barh", stacked=True, ax=ax, colormap="RdYlBu", width=.78)
base = (df["Tipo"].value_counts(normalize=True)["Correct"]) * 100
ax.axvline(base, ls="--", color="black", lw=1.3, label=f"base 'Correct' ({base:.0f}%)")
ax.set_title("Composición del razonamiento según la palabra usada en la explicación")
ax.set_xlabel("% de observaciones"); ax.set_ylabel("")
ax.legend(bbox_to_anchor=(1.01,1), loc="upper left", fontsize=8)
plt.tight_layout(); plt.show()
````

**Hallazgo 12 — El vocabulario sí discrimina, y de forma interpretable.**

La matriz TF-IDF tiene una densidad inferior al **0.2%**: extremadamente dispersa, lo esperable en
textos de 13 palabras de mediana.

El gráfico compara, para cada palabra, la composición de razonamientos frente a la tasa base de
`Correct`. Verbos de procedimiento matemático explícito (**divide**, **multiply**, **half**) se asocian
a razonamiento correcto muy por encima de la base. Marcadores de incertidumbre (**guess**, **think**)
se desplazan hacia `Neither`.

Esto confirma que un modelo basado en texto tiene señal real que aprovechar, y que los enfoques
clásicos (TF-IDF + regresión logística) son una línea base razonable antes de recurrir a
*transformers*.

| 6 | La longitud discrimina débilmente | Medianas 16 / 13 / 11 palabras | *Feature* auxiliar, no criterio único |
| 7 | Los outliers son válidos | Inspección manual de los extremos | Conservar; transformar con log |
| 8 | Multicolinealidad entre longitudes | ρ Spearman > 0.95 | Conservar solo `n_words` + normalizadas |
| 9 | Dificultad heterogénea entre preguntas | Acierto del 35.7% al 79.6% | `QuestionId` aporta contexto |
| 10 | **Concepciones erróneas atadas a la pregunta** | **32/35 exclusivas de 1 pregunta** | Condiciona toda la validación |
| 11 | La veracidad es casi determinista | 58/60 pares sin ambigüedad | Descomposición jerárquica del problema |
| 12 | El vocabulario discrimina | TF-IDF por tipo de razonamiento | Línea base clásica viable |

### 7.2 Conclusiones

**Sobre la calidad de los datos.** El conjunto está en buen estado para el EDA: no hay filas completas
idénticas, aunque existen 736 coincidencias al excluir `row_id` y 970 explicaciones repetidas. Sin un
identificador de estudiante no puede demostrarse independencia entre observaciones. El hallazgo de que
el 73% de los nulos era estructural evitó un error de
imputación que habría corrompido la mayor parte de los datos. La lección metodológica es que
diagnosticar el origen de una ausencia debe preceder siempre a la decisión de cómo tratarla.

**Sobre la dificultad real del problema.** El problema es considerablemente menos homogéneo de lo que
sugiere su formulación como clasificación de 65 clases. Se descompone en dos subproblemas de naturaleza
distinta: determinar la veracidad es casi trivial (58/60 pares deterministas), mientras que identificar
la concepción errónea concreta es donde reside toda la dificultad y donde el NLP es indispensable.

**Sobre la principal amenaza a la validez.** El hallazgo 10 es el más consecuente de todo el análisis.
Que el 91% de las concepciones erróneas sea exclusivo de una sola pregunta significa que **la estrategia
de validación importa más que la elección del modelo**. Una partición aleatoria producirá métricas
infladas que no se sostendrán ante preguntas nuevas; una partición agrupada por pregunta producirá
métricas artificialmente pésimas. Con 15 preguntas, ninguna de las dos opciones puras sirve.

### 7.3 Próximos pasos

1. **Validación doble.** Reportar sistemáticamente dos métricas: `StratifiedKFold` sobre la etiqueta
   compuesta (estima el rendimiento en las preguntas conocidas, comparable al *leaderboard* original) y
   `GroupKFold` por `QuestionId` (estima la generalización a preguntas nuevas). La distancia entre ambas
   es en sí misma un resultado que conviene medir y reportar.

2. **Modelo jerárquico en dos etapas.** Etapa 1: resolver `Veracidad` con una tabla de consulta sobre
   `(QuestionId, MC_Answer)`. Etapa 2: clasificar `Tipo` y la concepción errónea con un modelo de texto,
   restringiendo el espacio de salida a las concepciones observadas para esa pregunta — lo que reduce el
   problema de 65 clases a entre 2 y 4 por pregunta.

3. **Línea base antes que complejidad.** Establecer primero TF-IDF + regresión logística, ya justificada
   por el hallazgo 12. Solo entonces evaluar si un *transformer* afinado (DeBERTa, MathBERT) supera esa
   referencia lo suficiente para justificar su costo.

4. **Tratamiento de la cola larga.** Para las 24 clases con menos de 10 observaciones, evaluar
   agrupamiento en una categoría `Otros`, sobremuestreo, o aprovechar que MAP@3 permite tres intentos:
   incluir las clases raras plausibles en la 2.ª y 3.ª posición tiene un costo bajo y un beneficio
   potencial alto.

5. **Ingeniería de características.** Incorporar `n_words` (transformada con log), `digit_ratio`,
   `has_fraction` y la dificultad de la pregunta como variables complementarias al texto, descartando las
   redundantes identificadas en el hallazgo 8.
